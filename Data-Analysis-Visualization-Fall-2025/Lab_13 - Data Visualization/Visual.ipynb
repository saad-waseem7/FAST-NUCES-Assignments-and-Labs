```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

Load CSV Files (Section 1)



```python
sales = pd.read_csv("sales_data_sample.csv", encoding='latin1')
weather = pd.read_csv("weather.csv")
customers = pd.read_csv("customers.csv")
```

Task 2.1 — Line Plot (sales.csv)


```python
monthly_sales = sales.groupby('MONTH_ID')['SALES'].sum().reset_index()

monthly_sales_sorted = monthly_sales.sort_values(by='MONTH_ID')

plt.figure(figsize=(10, 6))
plt.plot(monthly_sales_sorted['MONTH_ID'], monthly_sales_sorted['SALES'], marker='o')
plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.title("Aggregated Monthly Sales Trend")
plt.xticks(monthly_sales_sorted['MONTH_ID'].unique())
plt.grid(True)
plt.show()
```

Task 2.2 — Compare Two Trends


```python
classic = sales[sales['PRODUCTLINE'] == "Classic Cars"]
moto = sales[sales['PRODUCTLINE'] == "Motorcycles"]

# Aggregate sales for Classic Cars by month
classic_monthly_sales = classic.groupby('MONTH_ID')['SALES'].sum().reset_index()
classic_monthly_sales_sorted = classic_monthly_sales.sort_values(by='MONTH_ID')

# Aggregate sales for Motorcycles by month
moto_monthly_sales = moto.groupby('MONTH_ID')['SALES'].sum().reset_index()
moto_monthly_sales_sorted = moto_monthly_sales.sort_values(by='MONTH_ID')

plt.figure(figsize=(12, 7))
plt.plot(classic_monthly_sales_sorted['MONTH_ID'], classic_monthly_sales_sorted['SALES'], label="Classic Cars", marker='o')
plt.plot(moto_monthly_sales_sorted['MONTH_ID'], moto_monthly_sales_sorted['SALES'], label="Motorcycles", marker='o')
plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.title("Aggregated Monthly Sales Trends of Two Product Categories")
plt.xticks(sales['MONTH_ID'].unique())
plt.grid(True)
plt.legend()
plt.show()
```
