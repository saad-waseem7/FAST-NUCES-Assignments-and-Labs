```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
#Q1
img = cv2.imread('lena.png', cv2.IMREAD_GRAYSCALE)
plt.imshow(img, cmap='gray')
plt.title('Grayscale Lena Image')
plt.axis('off')
plt.show()
```

```python
#Q2
plt.hist(img.ravel(), 256, [0,256])
plt.title('Histogram of Lena Image')
plt.show()
```

```python
#Q3
neg_img = 255 - img
plt.imshow(neg_img, cmap='gray')
plt.title('Negative of Lena Image')
plt.axis('off')
plt.show()

plt.hist(neg_img.ravel(), 256, [0,256])
plt.title('Histogram of Negative Image')
plt.show()
```

```python
#Q4
min_val = np.min(img)
max_val = np.max(img)
contrast_stretched = ((img - min_val) / (max_val - min_val) * 255).astype(np.uint8)

plt.imshow(contrast_stretched, cmap='gray')
plt.title('Contrast Stretched Image')
plt.axis('off')
plt.show()

plt.hist(contrast_stretched.ravel(), 256, [0,256])
plt.title('Histogram of Contrast Stretched Image')
plt.show()
```

```python
#Q5
mean_filtered = cv2.blur(img, (5,5))
median_filtered = cv2.medianBlur(img, 5)
gaussian_filtered = cv2.GaussianBlur(img, (5,5), 0)

plt.figure(figsize=(15,5))

plt.subplot(1,3,1)
plt.imshow(mean_filtered, cmap='gray')
plt.title('Mean Filtered')
plt.axis('off')

plt.subplot(1,3,2)
plt.imshow(median_filtered, cmap='gray')
plt.title('Median Filtered')
plt.axis('off')

plt.subplot(1,3,3)
plt.imshow(gaussian_filtered, cmap='gray')
plt.title('Gaussian Filtered')
plt.axis('off')

plt.show()
```

```python
plt.figure(figsize=(15,5))

plt.subplot(1,3,1)
plt.hist(mean_filtered.ravel(), 256, [0,256])
plt.title('Histogram of Mean Filtered')

plt.subplot(1,3,2)
plt.hist(median_filtered.ravel(), 256, [0,256])
plt.title('Histogram of Median Filtered')

plt.subplot(1,3,3)
plt.hist(gaussian_filtered.ravel(), 256, [0,256])
plt.title('Histogram of Gaussian Filtered')

plt.show()
```

```python
#Q6

_, th_simple = cv2.threshold(img, 127, 255, cv2.THRESH_BINARY)
plt.imshow(th_simple, cmap='gray')
plt.title('Simple Binary Threshold')
plt.axis('off')
plt.show()
plt.hist(th_simple.ravel(), 256, [0,256])
plt.title('Histogram - Simple Binary Threshold')
plt.show()

th_adaptive = cv2.adaptiveThreshold(img, 255, cv2.ADAPTIVE_THRESH_MEAN_C,
                                    cv2.THRESH_BINARY, 11, 2)
plt.imshow(th_adaptive, cmap='gray')
plt.title('Adaptive Thresholding')
plt.axis('off')
plt.show()
plt.hist(th_adaptive.ravel(), 256, [0,256])
plt.title('Histogram - Adaptive Threshold')
plt.show()

_, th_otsu = cv2.threshold(img, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)
plt.imshow(th_otsu, cmap='gray')
plt.title("Otsu's Thresholding")
plt.axis('off')
plt.show()
plt.hist(th_otsu.ravel(), 256, [0,256])
plt.title('Histogram - Otsu Threshold')
plt.show()
```
