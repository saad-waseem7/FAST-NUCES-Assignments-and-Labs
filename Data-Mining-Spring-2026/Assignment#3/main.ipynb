```python
# Read k-itemsets from a text file
# Each line in the file contains one itemset separated by spaces

def read_set(filename):
    itemsets = []

    file = open(filename, "r")
    for line in file:
        line = line.strip()
        if line != "":
            items = line.split()
            itemsets.append(items)
    file.close()

    return itemsets
```

```python
# Lk = Frequent k-itemsets

def sbst_found(subset, Lk):

    for itmst in Lk:
        if len(itmst) != len(subset):
            continue

        match = True

        for i in range(len(subset)):
            if subset[i] != itmst[i]:
                match = False
                break

        if match == True:
            return True

    return False

def infrq_sbst_fnd(candidate, Lk):

    k = len(candidate) - 1
    for i in range(len(candidate)):
        sbset = []

        for j in range(len(candidate)):
            if j != i:
                sbset.append(candidate[j])

        if sbst_found(sbset, Lk) == False:
            return True

    return False
```

```python
def apriori_algo(Lk):

    Ck = [] # Ck = Candidate k-itemsets
    for i in range(len(Lk)):
        for j in range(i+1, len(Lk)):

            l1 = Lk[i]
            l2 = Lk[j]
            join_possible = True

            for m in range(len(l1)-1):
                if l1[m] != l2[m]:
                    join_possible = False
                    break

            if join_possible:
                candidate = []
                for item in l1:
                    candidate.append(item)

                candidate.append(l2[-1])
                candidate.sort()

                if infrq_sbst_fnd(candidate, Lk) == False:
                    Ck.append(candidate)

    return Ck
```

```python
data = read_set("k_itemset.txt")

print("Frequent k-itemsets:")
for itmst in data:
    print(itmst)

res = apriori_algo(data)

print("\nCandidate (k+1)-itemsets:")
for k in res:
    print(k)
```
