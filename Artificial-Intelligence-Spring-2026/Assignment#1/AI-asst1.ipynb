```python
from collections import deque
```

```python
# Goal Test
def is_goal(state):
    _,dirt=state
    return dirt==(False, False)

# Successor Function
def get_successors(state):
    pos,(dirtA,dirtB) = state
    successors=[]

    # Clean
    if pos=="A" and dirtA:
        successors.append(("CLEAN",("A",(False,dirtB))))
    elif pos=="B" and dirtB:
        successors.append(("CLEAN",("B",(dirtA,False))))

    # Move Left
    if pos=="B":
        successors.append(("LEFT",("A",(dirtA,dirtB))))

    # Move Right
    if pos=="A":
        successors.append(("RIGHT",("B",(dirtA,dirtB))))

    return successors
```

```python
# Breadth First Search
def bfs(start):
    queue=deque([(start, [])])
    visited=set()
    states_generated=0

    while queue:
        state,path=queue.popleft()

        if state in visited:
            continue
        visited.add(state)

        if is_goal(state):
            return path,states_generated

        for action,next_state in get_successors(state):
            states_generated+=1
            queue.append((next_state,path+[(action, next_state)]))

    return None, states_generated

# Depth First Search
def dfs(start):
    stack=[(start,[])]
    visited=set()
    states_generated=0

    while stack:
        state,path=stack.pop()

        if state in visited:
            continue
        visited.add(state)

        if is_goal(state):
            return path,states_generated

        for action,next_state in get_successors(state):
            states_generated+=1
            stack.append((next_state,path+[(action, next_state)]))

    return None,states_generated

# Depth Limited Search 
def dls(state,path,depth,visited,states_generated):
    if is_goal(state):
        return path,states_generated

    if depth==0:
        return None,states_generated

    visited.add(state)

    for action,next_state in get_successors(state):
        states_generated+=1
        if next_state not in visited:
            result, states_generated=dls(
                next_state,
                path+[(action, next_state)],
                depth-1,
                visited,
                states_generated,
            )
            if result:
                return result,states_generated

    return None,states_generated

# Iterative Deepening Search
def ids(start,max_depth=10):
    total_generated=0

    for depth in range(max_depth):
        visited=set()
        result,generated=dls(start,[],depth,visited,0)
        total_generated+=generated

        if result:
            return result,total_generated

    return None,total_generated
```

```python
# Pretty Print
def show_result(name,path,states):
    print(f"\n{name} RESULT")
    print("-"*30)

    if not path:
        print("No solution found")
        return

    for step,(action,state) in enumerate(path,1):
        print(f"Step {step}: {action} -> {state}")

    print(f"Total moves: {len(path)}")
    print(f"States generated: {states}")

start_state=("A",(True,True))

bfs_path,bfs_states=bfs(start_state)
dfs_path,dfs_states=dfs(start_state)
ids_path,ids_states=ids(start_state)

show_result("BFS",bfs_path,bfs_states)
show_result("DFS",dfs_path,dfs_states)
show_result("IDS",ids_path,ids_states)
```

### Why BFS

- Explores level by level.
- Guaranteed to find the **optimal (shortest) solution** when all step costs are equal.
- Uses more memory than DFS.

In this problem: BFS is the best choice for optimal cleaning with minimum moves.

### Why DFS

- Goes deep first.
- Uses less memory than BFS.
- Not guaranteed to find the shortest path.
- Can get stuck exploring deep branches in larger problems.

In this problem: Works, but may produce a longer path than BFS.

### Why IDS

IDS combines the strengths of BFS and DFS:

- Memory efficient like DFS  
- Optimal like BFS (for unit cost)  
- Avoids deep infinite paths  

In this problem: IDS is often preferred in AI because it balances memory and optimality.

### Final Comparison

| Algorithm | Complete | Optimal | Memory |
|----------|----------|---------|--------|
| BFS | Yes | Yes | High |
| DFS | Yes (finite) | No | Low |
| IDS | Yes | Yes | Medium |

Conclusion:  
For the vacuum cleaner problem, BFS and IDS are optimal. IDS is generally preferred in AI because it uses less memory while still finding the shortest path.
